```python
import numpy as np
import pandas as pd
import os

parent_folder = os.path.dirname(os.getcwd())
part3_folder = os.path.join(parent_folder, "handoff_part3")
variants_folder = os.path.join(part3_folder, "variants")

# Load the raw test features
X_test_raw = pd.read_csv(
    os.path.join(variants_folder, "X_test_raw_all_sensors.csv")
)

# Load the true test labels
y_test = pd.read_csv(
    os.path.join(part3_folder, "y_test.csv")
).squeeze("columns")

# Confirm row counts match
print("Test feature rows:", len(X_test_raw))
print("Test label rows:", len(y_test))

assert len(X_test_raw) == len(y_test), "Feature and label row counts differ!"

# Use the saved model's own preprocessing and classifier
scores = saved_model["model"].predict_proba(
    saved_model["preprocessing"].transform(X_test_raw)
)[:, 1]

# Apply the saved decision threshold
threshold = saved_model["threshold"]
y_pred = (scores >= threshold).astype(int)

print("\nPrediction validation completed!")
print("Decision threshold:", threshold)
print("Predicted class counts:")
print(pd.Series(y_pred).value_counts().sort_index())
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    average_precision_score,
    confusion_matrix,
    classification_report
)

print("=== FINAL RANDOM FOREST EVALUATION ===\n")

print(f"Accuracy:  {accuracy_score(y_test, y_pred):.3f}")
print(f"Precision: {precision_score(y_test, y_pred, zero_division=0):.3f}")
print(f"Recall:    {recall_score(y_test, y_pred, zero_division=0):.3f}")
print(f"F1-score:  {f1_score(y_test, y_pred, zero_division=0):.3f}")
print(f"ROC-AUC:   {roc_auc_score(y_test, scores):.3f}")
print(f"Average Precision: {average_precision_score(y_test, scores):.3f}")

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred))

print("\nClassification Report:")
print(classification_report(
    y_test,
    y_pred,
    target_names=["Pass (0)", "Fail (1)"],
    zero_division=0
))
```

```python
import matplotlib.pyplot as plt
from sklearn.metrics import ConfusionMatrixDisplay, confusion_matrix

cm = confusion_matrix(y_test, y_pred, labels=[0, 1])

display_plot = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=["Pass", "Fail"]
)

fig, ax = plt.subplots(figsize=(7, 5))
display_plot.plot(ax=ax, values_format="d", cmap="Blues")
ax.set_title("Random Forest — Test Set Confusion Matrix")
plt.tight_layout()
plt.show()
```
